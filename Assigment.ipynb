1. What is Object-Oriented Programming (OOP)?

OOP is a way of writing code where we organize everything around objects. An object is like a real-world thing (car, dog, book). Each object has data (attributes) and actions (methods). This makes coding easier to understand, reuse, and maintain.

2. What is a class in OOP?

A class is like a blueprint. For example, the blueprint of a car tells you how to make a car. Similarly, a class tells you how to create objects.

3. What is an object in OOP?

An object is a real thing created from a class. If class is the blueprint of a car, then the car you drive is an object.

4. Difference between abstraction and encapsulation

Abstraction = hiding what we don’t need to see (like driving a car without knowing the engine’s working).

Encapsulation = keeping data and methods together in one place (like car engine parts sealed in one box).

5. What are dunder methods in Python?

Dunder = “double underscore”. These are special methods like __init__, __str__, etc., used to give objects special powers. Example: __add__ lets us use + on custom objects.

6. Explain inheritance in OOP

Inheritance means one class can reuse another class’s code. Example: A Dog class can inherit from Animal and get all its features without rewriting.

7. What is polymorphism in OOP?

Polymorphism means “many forms”. It allows the same function name to work in different ways depending on the object. Example: fly() means one thing for Sparrow and another for Penguin.

8. How is encapsulation achieved in Python?

We use private variables (_var or __var) and provide getter/setter methods to control access.

9. What is a constructor in Python?

A constructor (__init__) is a special method that runs automatically when we create an object. It sets up initial values.

10. What are class and static methods in Python?

Class method works with the class itself (not just objects) → defined using @classmethod.

Static method is a normal function inside a class that doesn’t depend on objects → defined using @staticmethod.

11. What is method overloading in Python?

Python doesn’t support overloading directly, but we can create methods that behave differently depending on how many/what type of arguments we pass.

12. What is method overriding in OOP?

When a child class writes its own version of a parent class method.

13. What is a property decorator in Python?

It allows us to use methods like attributes. Example: instead of calling obj.get_age(), we can just write obj.age.

14. Why is polymorphism important in OOP?

It makes code flexible and reusable. Same function can work with different types of objects.

15. What is an abstract class in Python?

An abstract class is a class that cannot be directly used. It only gives structure and forces child classes to implement certain methods.

```python
# 2. Write a program to create an abstract class Shape with a method area(). Derive classes Circle and Rectanglefrom it and implement the area() method in both.
from abc import ABC, abstractmethod

# Abstract class
class Shape(ABC):
    @abstractmethod
    def area(self):
        pass   # no implementation here, only a "contract"

# Derived class Circle
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius * self.radius

# Derived class Rectangle
class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width


# Testing
circle = Circle(5)
rectangle = Rectangle(4, 6)

print("Area of Circle:", circle.area())       # 78.5
print("Area of Rectangle:", rectangle.area()) # 24
```

```python
# 3. Implement a multi-level inheritance scenario where a class Vehicle has an attribute type. Derive a class Carand further derive a class ElectricCar that adds a battery attribute.

# Parent class
class Vehicle:
    def __init__(self, vehicle_type):
        self.type = vehicle_type

    def show_type(self):
        print("Vehicle Type:", self.type)


# Child class of Vehicle
class Car(Vehicle):
    def __init__(self, vehicle_type, brand):
        super().__init__(vehicle_type)   # call parent constructor
        self.brand = brand

    def show_car(self):
        print("Car Brand:", self.brand)


# Child class of Car (Grandchild of Vehicle)
class ElectricCar(Car):
    def __init__(self, vehicle_type, brand, battery_capacity):
        super().__init__(vehicle_type, brand)  # call Car constructor
        self.battery = battery_capacity

    def show_electric_car(self):
        print("Battery Capacity:", self.battery, "kWh")


# Testing
e_car = ElectricCar("Four Wheeler", "Tesla", 85)
e_car.show_type()          # from Vehicle
e_car.show_car()           # from Car
e_car.show_electric_car()  # from ElectricCar
```

```python
# Demonstrate polymorphism by creating a base class Bird with a method fly(). Create two derived classesSparrow and Penguin that override the fly() method.
# Base class
class Bird:
    def fly(self):
        print("Birds can fly (general behavior)")

# Derived class 1
class Sparrow(Bird):
    def fly(self):
        print("Sparrow flies high in the sky")

# Derived class 2
class Penguin(Bird):
    def fly(self):
        print("Penguin cannot fly, but swims very well")

# Polymorphism in action
birds = [Sparrow(), Penguin(), Bird()]

for b in birds:
    b.fly()   # same method name, different behavior
```

```python
# 5. Write a program to demonstrate encapsulation by creating a class BankAccount with private attributesbalance and methods to deposit, withdraw, and check balance.

class BankAccount:
    def __init__(self, initial_balance=0):
        # private attribute
        self.__balance = initial_balance

    # method to deposit money
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited: {amount}")
        else:
            print("Deposit amount must be positive")

    # method to withdraw money
    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            print(f"Withdrew: {amount}")
        else:
            print("Insufficient balance or invalid amount")

    # method to check balance
    def get_balance(self):
        print(f"Current Balance: {self.__balance}")


# ------------------------
# Using the class
acc = BankAccount(1000)
acc.deposit(500)      # Deposited: 500
acc.withdraw(300)     # Withdrew: 300
acc.get_balance()     # Current Balance: 1200

# Trying to access private variable directly
print(acc.__balance)   # ❌ Error (AttributeError: object has no attribute '__balance')
```

```python
# 6. Demonstrate runtime polymorphism using a method play() in a base class Instrument. Derive classes Guitar and Piano that implement their own version of play().
# Base class
class Instrument:
    def play(self):
        print("Instrument is playing (generic sound)")

# Derived class 1
class Guitar(Instrument):
    def play(self):
        print("Guitar is playing: Strum strum 🎸")

# Derived class 2
class Piano(Instrument):
    def play(self):
        print("Piano is playing: Ting ting 🎹")

# Runtime polymorphism in action
instruments = [Guitar(), Piano(), Instrument()]

for inst in instruments:
    inst.play()   # same method name -> different behavior at runtime
```

```python
# 7. Create a class MathOperations with a class method add_numbers() to add two numbers and a static method subtract_numbers() to subtract two numbers.
class MathOperations:
    # Class method (uses cls)
    @classmethod
    def add_numbers(cls, a, b):
        return a + b

    # Static method (no self, no cls)
    @staticmethod
    def subtract_numbers(a, b):
        return a - b


# Using the class
print(MathOperations.add_numbers(10, 5))
print(MathOperations.subtract_numbers(10, 5))

```

```python
# 8. Implement a class Person with a class method to count the total number of persons created.

class Person:
    # Class variable to keep count
    total_persons = 0

    def __init__(self, name):
        self.name = name
        Person.total_persons += 1   # increment count whenever a new object is created

    # Class method to get total persons
    @classmethod
    def get_total_persons(cls):
        return cls.total_persons


# -------------------
# Creating objects
p1 = Person("Aman")
p2 = Person("Ravi")
p3 = Person("Sita")

# Accessing class method
print("Total persons created:", Person.get_total_persons())
# Output: Total persons created: 3
```

```python
# 10. Demonstrate operator overloading by creating a class Vector and overriding the add method to add two vectors.
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    # Overloading +
    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __str__(self):
        return f"({self.x}, {self.y})"

# Usage
v1 = Vector(2, 3)
v2 = Vector(4, 5)
v3 = v1 + v2
print(v3)  # Output: (6, 8)

```

```python
# 12 .  Implement a class Student with attributes name and grades. Create a method average_grade() to compute the average of the grades.

class Student:
    def __init__(self, name, grades):
        self.name = name
        self.grades = grades

    def average_grade(self):
        return sum(self.grades) / len(self.grades) if self.grades else 0

# Usage
s = Student("Ravi", [80, 90, 70])
print("Average Grade:", s.average_grade())
```

```python
# 14. Create a class Employee with a method calculate_salary() that computes the salary based on hours worked and hourly rate. Create a derived class Manager that adds a bonus to the salary.

class Employee:
    def __init__(self, name, hours_worked, hourly_rate):
        self.name = name
        self.hours_worked = hours_worked
        self.hourly_rate = hourly_rate

    def calculate_salary(self):
        return self.hours_worked * self.hourly_rate

class Manager(Employee):
    def __init__(self, name, hours_worked, hourly_rate, bonus):
        super().__init__(name, hours_worked, hourly_rate)
        self.bonus = bonus

    def calculate_salary(self):
        return super().calculate_salary() + self.bonus

# Usage
e = Employee("Aman", 40, 200)
m = Manager("Ravi", 40, 200, 5000)
print("Employee Salary:", e.calculate_salary())  # 8000
print("Manager Salary:", m.calculate_salary())   # 13000
```

```python
# 16. Create a class Animal with an abstract method sound(). Create two derived classes Cow and Sheep that implement the sound() method.
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass

class Cow(Animal):
    def sound(self):
        print("Cow says: Moo 🐄")

class Sheep(Animal):
    def sound(self):
        print("Sheep says: Baa 🐑")


animals = [Cow(), Sheep()]
for a in animals:
    a.sound()
```

```python
# 17. Create a class Book with attributes title, author, and year_published. Add a method get_book_info() that returns a formatted string with the book's details.

class Book:
    def __init__(self, title, author, year_published):
        self.title = title
        self.author = author
        self.year_published = year_published

    def get_book_info(self):
        return f"'{self.title}' by {self.author}, published in {self.year_published}"


b = Book("Harry Potter", "J.K. Rowling", 1997)
print(b.get_book_info())
```

```python
# 18. Create a class House with attributes address and price. Create a derived class Mansion that adds an attribute number_of_rooms.

class House:
    def __init__(self, address, price):
        self.address = address
        self.price = price

class Mansion(House):
    def __init__(self, address, price, number_of_rooms):
        super().__init__(address, price)
        self.number_of_rooms = number_of_rooms


h = House("123 Main St", 500000)
m = Mansion("456 Luxury Ave", 5000000, 10)

print("House:", h.address, h.price)
print("Mansion:", m.address, m.price, "Rooms:", m.number_of_rooms)
```
